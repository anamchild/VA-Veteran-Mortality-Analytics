```python
# Clean and reshape the Recent VHA Veteran-All worksheet

years = [2018, 2019, 2020, 2021, 2022, 2023]

column_names = [
    "Rank",
    "Cause_of_Death",
    "Number",
    "Percent",
    "Crude_Rate",
    "Age_Adjusted_Rate",
    "YPLL",
    "YPLL_Percent"
]

recent_vha_yearly = []

for index, year in enumerate(years):

    start_col = index * 8
    end_col = start_col + 8

    # Extract one year's table
    yearly_table = recent_vha.iloc[4:, start_col:end_col].copy()
    yearly_table.columns = column_names

    # Add year and veteran group
    yearly_table["Year"] = year
    yearly_table["Veteran_Group"] = "Recent VHA"

    recent_vha_yearly.append(yearly_table)

# Combine all years
recent_vha_clean = pd.concat(recent_vha_yearly, ignore_index=True)

# Remove completely blank rows
recent_vha_clean = recent_vha_clean.dropna(
    how="all",
    subset=column_names
)

# Convert numeric columns
numeric_columns = [
    "Number",
    "Percent",
    "Crude_Rate",
    "Age_Adjusted_Rate",
    "YPLL",
    "YPLL_Percent"
]

for col in numeric_columns:

    recent_vha_clean[col] = (
        recent_vha_clean[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("--", "", regex=False)
        .str.strip()
    )

    recent_vha_clean[col] = pd.to_numeric(
        recent_vha_clean[col],
        errors="coerce"
    )

# Display first rows
recent_vha_clean.head(15)
```

## Years of Potential Life Lost (YPLL)

While mortality counts identify the leading causes of death, Years of Potential Life Lost (YPLL) provides a measure of premature mortality by estimating the number of years lost due to early death. Causes with the highest YPLL represent conditions that disproportionately affect younger veterans and therefore impose a greater burden on public health. Examining YPLL complements traditional mortality measures and provides additional insight into priorities for prevention and healthcare planning.

```python
# Top 10 causes of death by Years of Potential Life Lost (YPLL)

top10_ypll = (
    recent_vha_clean
    [
        (~recent_vha_clean["Cause_of_Death"].isin(["All cause", "Not ranked"])) &
        (recent_vha_clean["YPLL"].notna())
    ]
    .groupby("Cause_of_Death", as_index=False)["YPLL"]
    .sum()
    .sort_values("YPLL", ascending=False)
    .head(10)
)

top10_ypll
```

## Interpretation

The leading causes of Years of Potential Life Lost (YPLL) differed from the leading causes based solely on mortality counts. Malignant neoplasms (cancer) and diseases of the heart contributed the greatest burden of premature mortality, followed by unintentional injuries. Notably, intentional self-harm (suicide) ranked fourth in total YPLL despite accounting for fewer deaths than several chronic diseases. This indicates that suicide disproportionately affects younger veterans, resulting in a substantial loss of potential life years. These findings demonstrate the value of YPLL as a complementary measure for identifying priorities for prevention and healthcare planning among Recent Veterans Health Administration (VHA) users.

```python
def clean_veteran_sheet(df, veteran_group):

    years = [2018, 2019, 2020, 2021, 2022, 2023]

    column_names = [
        "Rank",
        "Cause_of_Death",
        "Number",
        "Percent",
        "Crude_Rate",
        "Age_Adjusted_Rate",
        "YPLL",
        "YPLL_Percent"
    ]

    yearly_tables = []

    for index, year in enumerate(years):

        start_col = index * 8
        end_col = start_col + 8

        yearly_table = df.iloc[4:, start_col:end_col].copy()
        yearly_table.columns = column_names

        yearly_table["Year"] = year
        yearly_table["Veteran_Group"] = veteran_group

        yearly_tables.append(yearly_table)

    cleaned_df = pd.concat(yearly_tables, ignore_index=True)

    cleaned_df = cleaned_df.dropna(
        how="all",
        subset=column_names
    )

    numeric_columns = [
        "Number",
        "Percent",
        "Crude_Rate",
        "Age_Adjusted_Rate",
        "YPLL",
        "YPLL_Percent"
    ]

    for col in numeric_columns:

        cleaned_df[col] = (
            cleaned_df[col]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("--", "", regex=False)
            .str.strip()
        )

        cleaned_df[col] = pd.to_numeric(
            cleaned_df[col],
            errors="coerce"
        )

    return cleaned_df
```

# Analysis 1 – Comparative Mortality Burden

## Research Question

How do the leading causes of death differ between Recent Veterans Health Administration (VHA) users and Other Veterans between 2018 and 2023?

### Analytical Approach

This analysis compares the cumulative number of deaths by cause for Recent Veterans Health Administration (VHA) users and Other Veterans across the six-year study period (2018–2023). Comparative population health analytics were used to aggregate mortality counts, rank the leading causes of death within each veteran population, and identify similarities and differences in mortality burden. The results are presented in both tabular and graphical formats to support interpretation and evidence-based healthcare planning.

```python
# Prepare cause-specific records for analysis

excluded_causes = [
    "All cause",
    "Not ranked"
]

cause_specific_data = combined_data[
    ~combined_data["Cause_of_Death"].isin(excluded_causes)
].copy()

print("Cause-specific dataset shape:", cause_specific_data.shape)
display(cause_specific_data.head())
```

```python
# Aggregate total deaths by cause and veteran group across 2018–2023

deaths_by_cause = (
    cause_specific_data
    .groupby(["Veteran_Group", "Cause_of_Death"], as_index=False)["Number"]
    .sum(min_count=1)
)

# Rank causes within each veteran group
deaths_by_cause["Death_Rank"] = (
    deaths_by_cause
    .groupby("Veteran_Group")["Number"]
    .rank(method="dense", ascending=False)
)

top_10_deaths = (
    deaths_by_cause
    .sort_values(["Veteran_Group", "Number"], ascending=[True, False])
    .groupby("Veteran_Group")
    .head(10)
)

display(top_10_deaths)
```

```python
# Create a side-by-side comparison table

top_causes_union = top_10_deaths["Cause_of_Death"].unique()

death_comparison_table = (
    deaths_by_cause[
        deaths_by_cause["Cause_of_Death"].isin(top_causes_union)
    ]
    .pivot(
        index="Cause_of_Death",
        columns="Veteran_Group",
        values="Number"
    )
    .sort_values(
        by=["Recent VHA", "Other Veteran"],
        ascending=False
    )
)

display(
    death_comparison_table.style.format(
        "{:,.0f}",
        na_rep="Suppressed"
    )
)
```

```python
import matplotlib.pyplot as plt

# Plot the leading causes of death for each veteran group

plot_data = (
    deaths_by_cause[
        deaths_by_cause["Cause_of_Death"].isin(top_causes_union)
    ]
    .pivot(
        index="Cause_of_Death",
        columns="Veteran_Group",
        values="Number"
    )
    .fillna(0)
)

plot_data["Combined_Total"] = plot_data.sum(axis=1)

plot_data = (
    plot_data
    .sort_values("Combined_Total", ascending=True)
    .drop(columns="Combined_Total")
)

ax = plot_data.plot(
    kind="barh",
    figsize=(12, 8)
)

ax.set_title(
    "Leading Causes of Death Among Recent VHA Users and Other Veterans, 2018–2023",
    fontsize=14,
    pad=15
)
ax.set_xlabel("Cumulative Number of Deaths")
ax.set_ylabel("Cause of Death")
ax.legend(title="Veteran Group")
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```

# Analysis 2 - Premature Mortality Burden (Years of Potential Life Lost)

## Research Question

To what extent does suicide contribute to premature mortality among Recent Veterans Health Administration (VHA) users and Other Veterans between 2018 and 2023, as measured by Years of Potential Life Lost (YPLL)?

### Analytical Approach

This analysis evaluates premature mortality using Years of Potential Life Lost (YPLL). Unlike mortality counts, YPLL emphasizes deaths occurring at younger ages and provides additional insight into the public health burden of preventable causes of death. Causes of death were ranked according to cumulative YPLL for each veteran population and compared across the study period.

```python
# Aggregate cumulative YPLL by cause and veteran group

ypll_by_cause = (
    cause_specific_data
    .groupby(
        ["Veteran_Group", "Cause_of_Death"],
        as_index=False
    )["YPLL"]
    .sum(min_count=1)
)

ypll_by_cause["YPLL_Rank"] = (
    ypll_by_cause
    .groupby("Veteran_Group")["YPLL"]
    .rank(
        method="dense",
        ascending=False
    )
)

top_10_ypll = (
    ypll_by_cause
    .sort_values(
        ["Veteran_Group", "YPLL"],
        ascending=[True, False]
    )
    .groupby("Veteran_Group")
    .head(10)
)

display(top_10_ypll)
```

```python
# Compare suicide deaths and YPLL

suicide_summary = (
    cause_specific_data[
        cause_specific_data["Cause_of_Death"] == "Intentional self-harm (suicide)"
    ]
    .groupby("Veteran_Group")
    .agg(
        Total_Deaths=("Number", "sum"),
        Total_YPLL=("YPLL", "sum")
    )
)

display(
    suicide_summary.style.format({
        "Total_Deaths": "{:,.0f}",
        "Total_YPLL": "{:,.0f}"
    })
)
```

```python
import matplotlib.pyplot as plt

plot_ypll = (
    ypll_by_cause[
        ypll_by_cause["Cause_of_Death"].isin(
            top_10_ypll["Cause_of_Death"].unique()
        )
    ]
    .pivot(
        index="Cause_of_Death",
        columns="Veteran_Group",
        values="YPLL"
    )
    .fillna(0)
)

plot_ypll["Combined_Total"] = plot_ypll.sum(axis=1)

plot_ypll = (
    plot_ypll
    .sort_values("Combined_Total", ascending=True)
    .drop(columns="Combined_Total")
)

ax = plot_ypll.plot(
    kind="barh",
    figsize=(12, 8)
)

ax.set_title(
    "Leading Causes of Premature Mortality (YPLL) Among Recent VHA Users and Other Veterans, 2018–2023",
    fontsize=14,
    pad=15
)
ax.set_xlabel("Cumulative Years of Potential Life Lost (YPLL)")
ax.set_ylabel("Cause of Death")
ax.legend(title="Veteran Group")
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```

### Key Findings

Cancer and heart disease accounted for the highest YPLL.

Accidents and suicide ranked much higher by YPLL than by death count.

Suicide ranked fourth in cumulative YPLL in both populations.

YPLL provided additional insight into premature mortality beyond traditional mortality counts.

### Interpretation

The Years of Potential Life Lost (YPLL) analysis revealed a different pattern from the mortality count analysis. Although heart disease and malignant neoplasms (cancer) remained the leading contributors to premature mortality, unintentional injuries and intentional self-harm (suicide) ranked substantially higher when measured by YPLL than when measured by mortality counts.

Most notably, intentional self-harm (suicide) ranked fourth in cumulative YPLL for both Recent Veterans Health Administration (VHA) users and Other Veterans. This indicates that, although suicide accounts for fewer deaths than many chronic diseases, it contributes disproportionately to premature mortality because deaths occur at younger ages, resulting in a greater loss of potential years of life.

These findings demonstrate that YPLL provides additional insight beyond traditional mortality counts and can help healthcare organizations identify conditions that have a substantial impact on premature mortality. This information can support evidence-based healthcare planning, resource allocation, and suicide prevention initiatives within the Veterans Health Administration.
